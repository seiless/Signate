```python
class CFG:
    TARGET = '含水率'
    DROP_COLS = ['sample number', 'species number']
```

```python
import pandas as pd
from read_data import load_datasets
from preprocess import train_df
```

```python
cols_to_analyze = [c for c in train_df.columns 
                   if c not in CFG.DROP_COLS + [CFG.TARGET]]
```

```python
df_sub = train_df[cols_to_analyze].select_dtypes(include=['number'])
def get_outlier_stats(df):
    results = []
    
    for col in df.columns:
        mean = df[col].mean()
        std = df[col].std()
        
        # 표준편차 2를 벗어나는 데이터 판별 (Z-score > 2)
        # |값 - 평균| > 2 * 표준편차
        outliers = df[(df[col] - mean).abs() > 2 * std]
        outlier_ratio = len(outliers) / len(df) * 100  # 백분율
        
        results.append({
            'Feature': col,
            'Mean': mean,
            'Std': std,
            'Min': df[col].min(),
            'Max': df[col].max(),
            'Outlier_Ratio(%)': round(outlier_ratio, 2)
        })
        
    return pd.DataFrame(results)

# 실행 및 결과 확인
summary_df = get_outlier_stats(df_sub)
```
