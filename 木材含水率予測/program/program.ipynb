```python
from preprocess import SpectralPreprocessor, load_datasets
import os
from learning import learn
from predicting import predict
```

```python
# パス設定（ユーザー定義のロジックを活用）
BASE_DIR = os.getcwd()
DATA_DIR = os.path.join(BASE_DIR, "data") 

# 1. データの読み込み
train_df, test_df = load_datasets()
```

```python
# 2. 前処理オブジェクトの生成
# ハイパーパラメータ最適化の際、window_length を変更して実験可能
preprocessor = SpectralPreprocessor(window_length=15)
X_train, y_train, sp_train = preprocessor.process_train_data(train_df)
X_test, sp_test, sample_ids = preprocessor.process_test_data(test_df)
```

```python
# 3. 学習 (種別ごとの PLS + LGBM アンサンブル)
# モデルファイルは program/ 内に保存される
bundle = learn(X_train, y_train, sp_train, use_ensemble=True)
```

```python
# 4. 予測 (単一サンプル予測ルールを遵守)
y_pred = predict(X_test, sp_test)
```

```python
import pandas as pd
submission = pd.DataFrame({
    "sample number": sample_ids,
    "含水率": y_pred
})
# 提出ファイルも data/ フォルダまたは root に保存可能
submission.to_csv("../output/submission.csv", index=False,header=False, encoding="sjis")
```
